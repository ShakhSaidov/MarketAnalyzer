```python
import pandas as pd
import numpy as np

# ==========================================
# EXACT PARAMETERS FROM YOUR ROW
# ==========================================
PARAMS = {
    'engulfing_type': 'wick',
    'session': 'US_Open',
    'trend_filter': 'All',
    'vol_filter': 'All',
    'min_streak': 1,
    'size_multi': 1.5,
    'hold_mins': 25,
    
    'norm_tp': 10.0,
    'norm_sl': 3.0,
    
    'inv_tp': 3.0,
    'inv_sl': 10.0,
}

def verify_discrepancy(p):
    print("Loading Parquet Data...")
    df = pd.read_parquet("../data/nq_1min.parquet")
    df.index = df.index.tz_localize('UTC').tz_convert('US/Pacific')
    
    df['hour'], df['minute'] = df.index.hour, df.index.minute
    cond = [
        ((df['hour'] == 0) | (df['hour'] == 1)),
        ((df['hour'] == 6) & (df['minute'] >= 30)) | (df['hour'] == 7) | ((df['hour'] == 8) & (df['minute'] < 30)),
        ((df['hour'] == 8) & (df['minute'] >= 30)) | (df['hour'] == 9) | (df['hour'] == 10) | ((df['hour'] == 11) & (df['minute'] < 30)),
        ((df['hour'] == 11) & (df['minute'] >= 30)) | (df['hour'] == 12)
    ]
    df['session'] = np.select(cond, ['London', 'US_Open', 'Midday', 'US_Close'], default='Other')
    df['total_size'] = df['high'] - df['low']
    df['color'] = np.where(df['close'] >= df['open'], 1, -1)
    df['prev_open'], df['prev_close'] = df['open'].shift(1), df['close'].shift(1)
    df['prev_high'], df['prev_low'] = df['high'].shift(1), df['low'].shift(1)
    df['prev_color'] = df['color'].shift(1)
    df['prev_total_size'] = df['total_size'].shift(1)
    df['streak'] = df['color'].groupby((df['color'] != df['prev_color']).cumsum()).cumsum().abs()
    df['prev_streak'] = df['streak'].shift(1)
    df['size_ratio'] = df['total_size'] / df['prev_total_size'].replace(0, 0.0001)

    bull_base = (df['prev_color'] == -1) & (df['color'] == 1) & (df['high'] > df['prev_high']) & (df['low'] < df['prev_low'])
    bear_base = (df['prev_color'] == 1) & (df['color'] == -1) & (df['high'] > df['prev_high']) & (df['low'] < df['prev_low'])

    ctx = (df['prev_streak'] >= p['min_streak']) & (df['size_ratio'] >= p['size_multi']) & (df['session'] == p['session'])
    bull_base, bear_base = bull_base & ctx, bear_base & ctx

    # Get valid signal indices
    signals = df[bull_base | bear_base].copy()
    
    high_arr = df['high'].values
    low_arr = df['low'].values
    close_arr = df['close'].values
    N_total = len(df)
    
    results = []
    
    print(f"Simulating Chronological Path for {len(signals)} trades...")
    
    for idx, row in signals.iterrows():
        i = df.index.get_loc(idx)
        is_bull = bull_base.iloc[i]
        ep = close_arr[i]
        
        # Normal Targets
        n_tp_lvl = ep + p['norm_tp'] if is_bull else ep - p['norm_tp']
        n_sl_lvl = ep - p['norm_sl'] if is_bull else ep + p['norm_sl']
        
        # Inverse Targets (Opposite direction)
        i_tp_lvl = ep - p['inv_tp'] if is_bull else ep + p['inv_tp']
        i_sl_lvl = ep + p['inv_sl'] if is_bull else ep - p['inv_sl']
        
        norm_outcome = "TIMEOUT"
        inv_outcome = "TIMEOUT"
        
        for step in range(1, p['hold_mins'] + 1):
            if i + step >= N_total: break
            h = high_arr[i + step]
            l = low_arr[i + step]
            
            # NORMAL CHECKS
            if norm_outcome == "TIMEOUT":
                n_tp_hit = (h >= n_tp_lvl) if is_bull else (l <= n_tp_lvl)
                n_sl_hit = (l <= n_sl_lvl) if is_bull else (h >= n_sl_lvl)
                if n_tp_hit and n_sl_hit: norm_outcome = "LOSS"
                elif n_tp_hit: norm_outcome = "WIN"
                elif n_sl_hit: norm_outcome = "LOSS"

            # INVERSE CHECKS
            if inv_outcome == "TIMEOUT":
                i_tp_hit = (l <= i_tp_lvl) if is_bull else (h >= i_tp_lvl)
                i_sl_hit = (h >= i_sl_lvl) if is_bull else (l <= i_sl_lvl)
                if i_tp_hit and i_sl_hit: inv_outcome = "LOSS"
                elif i_tp_hit: inv_outcome = "WIN"
                elif i_sl_hit: inv_outcome = "LOSS"
                
            if norm_outcome != "TIMEOUT" and inv_outcome != "TIMEOUT":
                break
                
        results.append({
            'Time': idx,
            'Signal': 'Bullish' if is_bull else 'Bearish',
            'Norm_Outcome': norm_outcome,
            'Inv_Outcome': inv_outcome
        })

    res_df = pd.DataFrame(results)
    
    # Calculate Stats
    total = len(res_df)
    n_wins = len(res_df[res_df['Norm_Outcome'] == 'WIN'])
    n_loss = len(res_df[res_df['Norm_Outcome'] == 'LOSS'])
    n_time = len(res_df[res_df['Norm_Outcome'] == 'TIMEOUT'])
    
    i_wins = len(res_df[res_df['Inv_Outcome'] == 'WIN'])
    i_loss = len(res_df[res_df['Inv_Outcome'] == 'LOSS'])
    i_time = len(res_df[res_df['Inv_Outcome'] == 'TIMEOUT'])
    
    double_loss = len(res_df[(res_df['Norm_Outcome'] == 'LOSS') & (res_df['Inv_Outcome'] == 'LOSS')])

    print("\n" + "="*50)
    print("📊 VERIFICATION REPORT 📊")
    print("="*50)
    print(f"Total Trades Evaluated: {total:,}")
    print("\n--- SCRIPT A (NORMAL TRADE) ---")
    print(f"Normal Win%:     {(n_wins/total)*100:.1f}%")
    print(f"Normal Loss%:    {(n_loss/total)*100:.1f}%")
    print(f"Normal Timeout%: {(n_time/total)*100:.1f}%")
    
    print("\n--- SCRIPT B (INVERSE TRADE) ---")
    print(f"Inverse Win%:    {(i_wins/total)*100:.1f}%")
    print(f"Inverse Loss%:   {(i_loss/total)*100:.1f}%")
    print(f"Inverse Timeout%:{(i_time/total)*100:.1f}%")
    
    print("\n--- THE DISCREPANCY EXPLAINED ---")
    print(f"Timeouts Match?      {'Yes' if n_time == i_time else 'No'}")
    print(f"Double Losses (Tie-breaks): {double_loss:,} trades ({(double_loss/total)*100:.1f}%)")
    print(f"\nMath Check: Normal Loss% ({(n_loss/total)*100:.1f}%) - Double Loss% ({(double_loss/total)*100:.1f}%) = Inverse Win% ({(i_wins/total)*100:.1f}%)")
    print("="*50)

    # Save to CSV for manual check
    res_df.to_csv("verify_side_by_side.csv", index=False)
    print("Saved full side-by-side list to 'verify_side_by_side.csv'")

if __name__ == "__main__":
    verify_discrepancy(PARAMS)
```
