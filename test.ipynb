```python
import pandas as pd

pd.set_option("display.max_columns", None)
pd.set_option("display.max_rows", None)

# Load data
df = pd.read_parquet("./data/nq_1min.parquet")

# Existing timestamps represent UTC, but are timezone-naive
df.index = df.index.tz_localize("UTC")

# Convert UTC -> US Eastern Time
df.index = df.index.tz_convert("America/New_York")

# Check the first rows
print(df.head())

# Check rows starting from January 3, 2019 at 09:00 Eastern Time
start = pd.Timestamp("2019-01-07 09:00:00", tz="America/New_York")

df.loc[start:].head(100)
```
